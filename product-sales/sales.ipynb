```python
import pandas as pd
```

```python
df = pd.read_csv('product.csv')
```

```python
df.info()
```

```python
df.head()
```

```python
# Calculate the total revenue generated across all sales.
total_revenue = df['Total Revenue'].sum()
print(f"Total Revenue: ${total_revenue:,.2f}")
```

```python
# Group the data by Region and find the total Units Sold for each region.
units_by_region = df.groupby('Region')['Units Sold'].sum()
units_by_region
```

```python
# Identify the date with the highest Total Revenue.
max_date = df.loc[df['Total Revenue'].idxmax(), 'Date']
max_date
```

```python
# Filter and display all records where the Product is "Widget b"
widget_b_sales = df[df['Product'] == 'Widget B']
widget_b_sales
```

```python
# Calculate the average Unit Price for each Product.
avg_price_per_product = df.groupby('Product')['Unit Price'].mean()
print(avg_price_per_product)
```

```python
#Add a new column called Revenue per Unit calculated as Total Revenue / Units Sold.
df['Revenue per Unit'] = df['Total Revenue'] / df['Units Sold']
```
